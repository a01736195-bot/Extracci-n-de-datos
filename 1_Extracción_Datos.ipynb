```python
#Carga desde un archivo .xlsx sin indice
df= pd.read_excel('Citas_Digital.xlsx')
df
```

```python
#Verificamos información del DataFrame
df.info()
```

```python
#Filtro por filas
#filtro1= df.iloc[3:10,: ]           #Filas consecutivas
#filtro1=df.iloc[ [8, 24, 100] , : ]  #Filas no consecutivas
#filtro1= df.iloc[ [0] + list(range(10, 14)) + [50] , [0, 1] ]  #Filas combinadas
filtro1= df.iloc[0:673,: ]
filtro1.info()
```

```python
#Filtro por columnas
#filtro2=df.iloc[ : , 0:3]       #Columnas consecutivas
#filtro2=df.iloc[ : , [1,4,6]]  #Columnas no consecutivas
#filtro2= df.iloc[ :, list(range(1, 6)) + 9]  #Filas combinadas
filtro2 = df.iloc[ : , list(range(0, 3))+ list(range(4, 15))] #En los rangos se debe colocar +1 del limite superior
filtro2.info()
```

```python
#Convertir archivo filtrado a CSV
filtro3.to_csv("Citas_Digital_Filtrado.csv", index=False)
```

**FILTROS BASICOS PARA VARIABLES CUANTITATIVAS**

```python
#Filtro por comparacion
filtro8=df_filtrado[df_filtrado["Potencial de compra"] < 0.5]
filtro8
```

```python
#Conversión de tipo de dato de columna de tipo String a Float
df_filtrado['Potencial de compra']= df_filtrado['Potencial de compra'].astype(float)
```

```python
#Reemplazo de datos de una columna 
df_filtrado['Potencial de compra']=df_filtrado['Potencial de compra'].str.replace("Ventas ","0", regex=False)
```

```python
df_filtrado['Potencial de compra']=df_filtrado['Potencial de compra'].str.replace("%","", regex=False)
```

```python
df_filtrado['Potencial de compra']=df_filtrado['Potencial de compra'].str.replace("'","", regex=False)
```

```python
df_filtrado['Potencial de compra']=df_filtrado['Potencial de compra'].str.replace("&","", regex=False)
```

**FILTROS BASICOS CON CONETORES LOGICOS**

```python
#Filtro Y
filtro9=df_filtrado[(df_filtrado["Potencial de compra"] >= 0.9) & (df_filtrado["Asesor Asignado"] == "Aurelio")]
filtro9
```

```python
#Filtro o
filtro10=df_filtrado[(df_filtrado["Potencial de compra"] < 0.2) | (df_filtrado["BDC "] == "Saul")]
filtro10
```

```python
#Filtro not
filtro11=df_filtrado[~(df_filtrado["BDC "] == "Antonio")]
filtro11
```

```python
#Convertir archivo filtrado a CSV
filtro11.to_csv("filtro11.csv", index=False)
```
