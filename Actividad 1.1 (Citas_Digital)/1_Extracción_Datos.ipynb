```python
# ═══ CELDA 1 — Leer el Excel ═══
import pandas as pd
df = pd.read_excel('Citas_Digital.xlsx')

# ═══ CELDA 2 — Paso 1: filtro por filas (0 a 673) ═══
df = df.iloc[0:674]
df

# ═══ CELDA 3 — Paso 2: filtro por columnas (0,1,2,4,5,6,7,8,9) ═══
df = df.iloc[:, [0,1,2,4,5,6,7,8,9]]
df

# ═══ CELDA 4 — Paso 3: limpiar "Potencial de compra" y convertir a float ═══
df['Potencial de compra'] = df['Potencial de compra'].astype(str)
df['Potencial de compra'] = df['Potencial de compra'].str.replace("Ventas ", "0", regex=False)
df['Potencial de compra'] = df['Potencial de compra'].str.replace("%", "", regex=False)
df['Potencial de compra'] = df['Potencial de compra'].str.replace("'", "", regex=False)
df['Potencial de compra'] = df['Potencial de compra'].str.replace("&", "", regex=False)
df['Potencial de compra'] = df['Potencial de compra'].astype(float)
df['Potencial de compra']

# ═══ CELDA 5 — Paso 4: exportar CSV general ═══
df.to_csv('Citas_Digital_General.csv', index=False)

# ═══ CELDA 6 — Paso 5: leer el CSV general y convertir Fecha ═══
df_general = pd.read_csv('Citas_Digital_General.csv')
df_general['Fecha'] = pd.to_datetime(df_general['Fecha'])

# ═══ CELDA 7 — Paso 5: un CSV por mes (abril 2025 a julio 2026) ═══
meses = [(4,2025,'Abril_2025'),(5,2025,'Mayo_2025'),(6,2025,'Junio_2025'),(7,2025,'Julio_2025'),
         (8,2025,'Agosto_2025'),(9,2025,'Septiembre_2025'),(10,2025,'Octubre_2025'),(11,2025,'Noviembre_2025'),
         (12,2025,'Diciembre_2025'),(1,2026,'Enero_2026'),(2,2026,'Febrero_2026'),(3,2026,'Marzo_2026'),
         (4,2026,'Abril_2026'),(5,2026,'Mayo_2026'),(6,2026,'Junio_2026'),(7,2026,'Julio_2026')]
for mes, anio, nombre in meses:
    filtro = df_general[(df_general['Fecha'].dt.month == mes) & (df_general['Fecha'].dt.year == anio)]
    filtro.to_csv(nombre + '.csv', index=False)
```
